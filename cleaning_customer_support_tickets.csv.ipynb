## BUSINESS ANALYSTICS ## 

### Create Resolution Duration Metric ###

```python
df_cpy1['resolution_time_hours'] = (df_cpy1['time_to_resolution']-df_cpy1['first_response_time']).dt.total_seconds()/3600
```

```python
df_cpy1.head()
```

```python
df_cpy1[df_cpy1['resolution_time_hours']<0]
```

```python
negetive_df_cpy1 = df_cpy1[df_cpy1['resolution_time_hours']<0]

negetive_df_cpy1[
    ['ticket_id','first_response_time','time_to_resolution','ticket_status']
].head(20)
```

```python
negetive_df_cpy1
```

### Create Data Quality Flag ###

```python
df_cpy1['data_quality_flag'] = df_cpy1['resolution_time_hours'].apply(
    lambda x:'Invalid Duration'
    if pd.notnull(x) and x<0
    else 'Valid'
)
```

```python
df_cpy1.head()
```

### Create Clean Resolution Metric ###

```python
df_cpy1.loc[df_cpy1['resolution_time_hours']<0,'resolution_time_hours'] = None
```
